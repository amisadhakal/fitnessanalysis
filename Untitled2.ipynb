```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
```

```python
df = pd.read_csv("clean_dataset.csv")
print(df.columns)
df.head()
```

```python
X = df[[
    'steps',
    'calories_burned',
    'distance_km',
    'active_minutes',
    'sleep_hours',
    'heart_rate_avg'
]]
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
inertia = []

for k in range(1, 10):
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X_scaled)
    inertia.append(kmeans.inertia_)

plt.plot(range(1, 10), inertia, marker='o')
plt.xlabel("K") 
plt.ylabel("Inertia")
plt.title("Elbow Method")
plt.show()
```

```python
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
df['Cluster'] = kmeans.fit_predict(X_scaled) 
```

```python
print(df[['steps','active_minutes','sleep_hours','Cluster']].head())
```

```python
cluster_summary = df.groupby('Cluster').mean(numeric_only=True)
print(cluster_summary)
```

```python
def label_cluster(row):
    if row['distance_km'] > 10:
        return "High Activity Users"
    elif row['calories_burned'] > 3000:
        return "High Burn Users"
    else:
        return "Moderate Users"

df['Cluster_Label'] = df.apply(label_cluster, axis=1)
```

```python
import seaborn as sns
```

```python
sns.scatterplot(
    data=df,
    x='steps',
    y='calories_burned',
    hue='Cluster',
    palette='viridis'
)

plt.title("KMeans Clustering of Fitness Data")
plt.show()
```
